## Question F11: Functions and error handling

Write a function called `say_hello` that returns a greeting message: "Hey there, guest_name!"

- When the function is called without any parameter, the displayed `guest_name` will default to "Jordan".
- Otherwise, the `guest_name` is the string value passed.
  
Ensure the function handles the following error cases:
- If None is passed, the function raises a `ValueError` with the message: "guest_name cannot be None".
- If an empty string is passed, the function raises a `ValueError` with the message: "guest_name cannot be an empty string".
- If a non-string value is passed (e.g., an integer), the function raises a `TypeError`.

```python
def say_hello(guest_name="Jordan"):
    """
    Say hello to the guest with a personalised message.

    Parameters:
    - guest_name (str): The name of the guest (default is "Jordan").
   
    Returns:
    str: A greeting message.
    """
    # YOUR CODE HERE
    if guest_name is None:
        raise ValueError("guest_name cannot be None")
    
    if not isinstance(guest_name, str):
        raise TypeError("guest_name must be a string")
    
    if guest_name == "":
        raise ValueError("guest_name cannot be an empty string")
    
    return f"Hey there, {guest_name}!"
```

```python
"""Check that say_hello returns the correct output for several inputs"""

# Test Case 1
assert say_hello() == "Hey there, Jordan!"

# Test Case 2
assert say_hello("Mia") == "Hey there, Mia!"

# Test Case 3
assert say_hello("Liam") == "Hey there, Liam!"

# Test Case 4
assert say_hello(guest_name="Zara") == "Hey there, Zara!"
```

```python
"""Check that say_hello raises an error for invalid inputs"""

# Test Case 5: Providing None as the guest_name
try:
    say_hello(None)
except ValueError as e:
    assert str(e) == "guest_name cannot be None"
else:
    raise AssertionError

# Test Case 6: Providing an empty string as the guest_name
try:
    say_hello("")
except ValueError as e:
    assert str(e) == "guest_name cannot be an empty string"
else:
    raise AssertionError
    
# Test Case 7: Providing an invalid type for guest_name
try:
    say_hello(123)
except TypeError as e:
    pass
else:
    raise AssertionError
```

## Question F12: Nested loop - Cumulative Addition Grid

Write a function called `create_cumulative_grid` that takes an integer `size` as input and returns a 2D list ("list of lists") representing a cumulative addition grid up to `size × size`.

Each inner list should represent a row of the grid, and each element in the inner list should be the sum of the corresponding row and column indices instead of their product.

For example, if size = 3, the output should be:

[[2, 3, 4],  
 [3, 4, 5],  
 [4, 5, 6]]

(Note: Each value is computed as row_index + col_index, starting from 1.)

```python
def create_cumulative_grid(size):
    """
    Generate a 2D list representing a cumulative addition grid up to size × size.

    Parameters:
    - size (int): The size of the grid.

    Returns:
    list of lists: A cumulative sum grid of dimensions size × size.
    """
    # YOUR CODE HERE
    grid = []
    
    for row_index in range(1, size + 1):
        current_row = []
        for col_index in range(1, size + 1):
            value = row_index + col_index
            current_row.append(value)
        grid.append(current_row)
    
    return grid
    
    
```

```python
"""Check that create_cumulative_grid returns the correct output for several inputs"""

# Test Case 1
assert create_cumulative_grid(2) == [[2, 3], [3, 4]]

# Test Case 2
assert create_cumulative_grid(3) == [[2, 3, 4], [3, 4, 5], [4, 5, 6]]

# Test Case 3
assert create_cumulative_grid(4) == [[2, 3, 4, 5], [3, 4, 5, 6], [4, 5, 6, 7], [5, 6, 7, 8]]
```

## Question F15: Regression evaluation metrics

Write a function called `regression_rmse` to calculate <b>root mean squared error (RMSE)</b>. Follow the detailed steps as given below, steps (1-3) are same as in question F14. 

### Steps: 

1. <b>Import</b> required libraries.
2. <b>Read</b> the file `melb_data.csv` into a dataframe named `test_df`.
3. <b>Update</b> `test_df`: remove rows with missing values.
4. <b>Filter the dataframe</b> to include only properties where the number of rooms is greater than 4. Store the filtered data in a new dataframe named `df_filtered_by_rooms`.
5. The <b>feature variable(s)</b> will be provided as a list when calling the function. Use this list to select the predictor variables.
6. Use <b>Price</b> as the <b>target variable</b>.
7. Use the `sklearn` library to:
    - <b>Split the dataset</b> into training and testing sets (test_size=0.20, random_state=0).
    - Create a <b>linear regression model</b> and <b>fit</b> it on the training set.
    - <b>Make predictions</b> on the test set.
  
8. Compute the <b>Root Mean Squared Error (RMSE).</b>
9. Return the RMSE value.

Note: The test cases include a `round` function to allow for minor discrepancies in the decimal digits. Your code does not need to perform any rounding.

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn import metrics
import numpy as np
import pandas as pd

def regression_rmse(feature_variable_list):
    """ 
    Perform a regression task using a linear regression model and calculate the Root Mean Squared Error (RMSE).

    Parameters:
    - feature_variable_list (list): List of column names to be used as predictor variables.

    Returns:
    float: The RMSE value for the regression model.
    """
    
    # YOUR CODE HERE
    test_df = pd.read_csv("melb_data.csv")
    test_df = test_df.dropna()
    df_filtered_by_rooms = test_df[test_df['Rooms'] > 4]
    X = df_filtered_by_rooms[feature_variable_list]
    y = df_filtered_by_rooms['Price']
    X_train, X_test, y_train, y_test = train_test_split(
        X, y, 
        test_size=0.20, 
        random_state=0
    )
    model = LinearRegression()
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    rmse = np.sqrt(metrics.mean_squared_error(y_test, y_pred))
    return rmse
```

```python
"""Check that regression_rmse returns the correct output for several inputs"""

# Test Case 1
rmse = regression_rmse(['Rooms', 'Distance_from_CBD', 'Bathroom', 'Car', 'Landsize', 'BuildingArea', 'Propertycount'])
assert round(rmse, 2) == 896277.59

# Test Case 2
rmse = regression_rmse(['Rooms', 'Distance_from_CBD', 'Landsize'])
assert round(rmse, 2) == 697019.18

# Test Case 3
rmse = regression_rmse(['BuildingArea'])
assert round(rmse, 2) == 1576733.45

# Test Case 4
rmse = regression_rmse(['Landsize', 'Rooms', 'BuildingArea'])
assert round(rmse, 2) == 1416205.79

# Test Case 5
rmse = regression_rmse(['Bathroom'])
assert round(rmse, 2) == 816721.91
```
