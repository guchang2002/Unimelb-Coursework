## Basic Introduction

You will be introduced to spatial data in this assignment. 

<b> Spatial data </b> observations focus on locations. Every house, city, tree, car any object has its own unique latitude and longitude coordinates. 

The two primary types of spatial data are <b> vector and raster data </b> in GIS. In this Assignment 2 model (tutorials and assignment), you will explore vector data to the fullest. 

Vector graphics consist of vertices and paths, and the fundamental symbol categories for <b>vector data are points, lines and polygons.</b>

<b>Vector points</b> are essentially coordinates in XY format, typically represented by latitude and logitude within a spatial reference framework. Points are used when features are too small to be illustrated as polygons. For example, Maps in global scale often use points to display cities, since boundary lines are not visible at such a vast level of observation. 

<b>Vector lines</b> connect each vertex with paths. Lines usually represent features that are linear in nature. For example, maps show rivers, roads, pipelines as vector lines. 

<b>Polygons</b> connect vertices and close the path. When you join a set of vertices in a particular order and close it, this is now a vector polygon feature. Cartographers use polygons to show boundaries and they all have an area. For example, a building footprint has square footage.

## Question F3 

Write a function called `wgs84_to_epsg7855` that takes coordinates in WGS84 and projects them to GDA 2020 MGA Zone 55 (EPSG: 7855), returning the projected coordinates. 

<i>
Note: 

Coordinates (longitude, latitude) in Australia are typically in the World Geodetic System 1984 (WGS84) or EPSG: 4326. You need to project these coordinate to GDA 2020 MGA Zone 55 (EPSG: 7855). The GDA 2020 MGA Zone 55 projected coordinate system is appropriate for parts of Australia that lie between 144°E and 150°E.

Refer to Practical 2.1 to learn more about coordinate reference systems. </i>

```python
import pyproj
from shapely.geometry import Point
from shapely.ops import transform

def wgs84_to_epsg7855(long, lat): 
    """
    Project coordinates from WGS84 to GDA 2020 MGA Zone 55 (EPSG 7855).

    Parameters:
    - long: float, longitude coordinate in WGS84.
    - lat: float, latitude coordinate in WGS84.

    Return:
    - projected_longitude: float, longitude coordinate in GDA 2020 MGA Zone 55.
    - projected_latitude: float, latitude coordinate in GDA 2020 MGA Zone 55.
    """
    # creating a Point object by passing the longitude and latitude of the city to the Point() function.   
    coord = Point(long, lat)
    
    # YOUR CODE HERE
    wgs84 = pyproj.CRS("EPSG:4326")  # WGS84
    epsg7855 = pyproj.CRS("EPSG:7855")  # GDA 2020 MGA Zone 55
    transformer = pyproj.Transformer.from_crs(wgs84, epsg7855, always_xy=True)
    projected_coord = transform(transformer.transform, coord)
    return projected_coord.x, projected_coord.y
    
```

```python
"""Check that wgs84_to_epsg7855 returns the correct output for several inputs"""
# Test Case 1
x,y = wgs84_to_epsg7855(144.96, -37.8)
assert (round(x,2), round(y,2)) == (320398.59, 5814414.88)

# Test Case 2
x,y = wgs84_to_epsg7855(147.35005058424463, -35.13235260730235)
assert (round(x,2), round(y,2)) == (531891.24, 6112223.31)

# Test Case 3
x,y = wgs84_to_epsg7855(149.13596713397143, -35.279551459965944)
assert (round(x, 2), round(y,2)) == (694260.43, 6093863.1)
```

## Question F4

Write a function called `calculate_distance` that calculates the straight-line distance in metres from Melbourne to another city(with longitude and latitude provided), using the MGA Zone 55 coordinate system.

```python
import pyproj
from shapely.geometry import Point
from shapely.ops import transform

def calculate_distance(longitude, latitude):
    """
    Calculate the straight-line distance in meters from Melbourne to another city,
    using the MGA Zone 55 coordinate system. 
    
    Steps:
        1. Create Point objects for both Melbourne and the other city using Point() function. 
        2. Project both points from WGS84 to GDA2020 MGA Zone 55 (EPSG: 7855).
        3. Calculate the distance between the projected points. 
        4. Return the distance in metres. 

    Parameters:
    - longitude: float, longitude of the other city (in degrees, WGS84).
    - latitude: float, latitude of the other city (in degrees, WGS84).

    Return:
    - float: The straight-line distance in meters from Melbourne to the specified city.
    """
        
    # Create Point objects using WGS84 coordinates (longitude, latitude)
    melbourne = Point (144.9631,-37.8136)   # Coordinates of Melbourne
    city = Point (longitude, latitude) 
    

    # YOUR CODE HERE
    transformer = pyproj.Transformer.from_crs("EPSG:4326", "EPSG:7855", always_xy=True)
    melbourne_projected = transform(transformer.transform, melbourne)
    city_projected = transform(transformer.transform, city)
    distance = melbourne_projected.distance(city_projected)
    return distance
```

```python
"""Check that calculate_distance returns the correct output for several inputs"""
# Test Case 1- City 1 (Sydney)
assert round(calculate_distance(151.2093, -33.8688), 2) == 713906.34

# Test Case 2- City 2 (Adelaide)
assert round(calculate_distance(138.59863, -34.92866), 2) == 656294.76 

# Test Case 3 City 3- (Canberra)
assert round(calculate_distance(149.1290983846947, -35.277878774638616), 2) == 467039.81
```

## Question F5

Write a function called `calculate_area` that reads the file "Western Victoria agriculture.gpkg', which represents the LGAs within Western Victoria. The function should calculate an return the area in square kilometres for the LGA specified in the function call. 

Steps(hints):
1. Use GeoPandas to read the file and store it in a GeoDataFrame. (The file path has been already provided for consistency.)
2. Convert the coordinate reference system (CRS) from <b>GDA94</b> to <b>GDA 2020 MGA Zone 55 (EPSG: 7855)</b>  using `to_crs()` function.
3. Calculate the area of the specified LGA.
4. As the calculated area will be in square metres, convert it to square kilometres.
5. Return the calculated area in square kilometres.


```python
import geopandas as gpd
def calculate_area(LGA_NAME):
    """
    Read the GeoPackage file and calculate the area, in square kilometers, for the specified Local Government Area (LGA).

    Parameters:
    -LGA_NAME: str, the name of the LGA for which the area is to be calculated.
    
    Returns
    - float: The calculated area of the specified LGA in square kilometres.
    """
    # Read the GeoPackage containing LGAs in Western Victoria
    geotables_polygons = gpd.read_file("data/Western Victoria agriculture.gpkg")
    
    # YOUR CODE HERE
    geotables_polygons = geotables_polygons.to_crs("EPSG:7855")
    selected_lga = geotables_polygons[geotables_polygons['LGA_NAME'] == LGA_NAME]
    area_sqm = selected_lga.geometry.area.iloc[0]
    area_sqkm = area_sqm / 1000000
    return area_sqkm
```
